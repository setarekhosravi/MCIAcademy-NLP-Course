```python
def read_file(path_to_file):
  
  data = []

  with open(path_to_file, 'r') as ptr:
      for line in ptr:
        tmp = line.strip() 
        tmp = tmp.replace('؟', '')
        tmp = tmp.replace('.', '')
        tmp = tmp.replace('!', '')
        tmp = tmp.replace('،', '')
        data.append(tmp)
  
  return data
```




### We start with:

*   Unigram LM





```python
# calculate unigram probabilities for each word (w_i) and store it

def calc_unigram_probs(corpus):
  
  word_count = {}
  total_words = 0

  for doc in corpus:
    tmp = doc.split(' ')
    total_words += len(tmp)

    for word in tmp:
      if (not (word in word_count)):
        word_count[word] = 0
      word_count[word] = word_count[word] + 1

  return word_count, total_words
```

```python
word_count, total_words = calc_unigram_probs(sentences)
```

```python
def get_unigram_prob(word):

  # prob =  math.log((word_count[word]/total_words))
  prob = (word_count[word]/total_words)
  return prob
```

```python
# Bigrams are extracted from input corpus

def collect_bigrams(corpus):
  
  word_bigrams = {}
  
  for doc in corpus:
    tmp = doc.split(' ')

    for i in range(1, len(tmp)):
      word = tmp[i]
      prev_word = tmp[i-1]
      
      if not (word in word_bigrams):
        word_bigrams[word] = []

      word_bigrams[word].append(prev_word)

  return word_bigrams
```

```python
word_bigrams = collect_bigrams(sentences)
```

```python
print(word_bigrams['محکم'])
print(len(word_bigrams['محکم']))
```

```python
# The bigram LM probability of the given bigram is calculated

def get_bigram_prob(word, prev_word):
  
  bigrams = word_bigrams[word]
  count = 0
  
  for elem in bigrams:
    if elem == prev_word:
      count += 1

  N = word_count[prev_word]
  prob = count / N

  return prob
```
